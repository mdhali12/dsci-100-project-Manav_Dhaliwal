```python
import altair as alt

plot_1 = alt.Chart(tidied_df).mark_circle(size=60).encode(
    x=alt.X('reg_wins_pct:Q', title='Regular Season Win Percentage (%)'),
    y=alt.Y('total_wins:Q', title='Total Wins'),
    tooltip=['school', 'reg_wins_pct', 'conf_rank', 'seed', 'total_wins']
).properties(
    title='Total Wins vs. Regular Season Win Percentage',
    width=400,
    height=300
)

plot_1
```

```python
plot2 = alt.Chart(tidied_df).mark_bar().encode(
    alt.X("seed:N", title="Tournament Seed", sort="ascending"),
    alt.Y("count()", title="Count / Proportion"),
    color="seed:N"
).properties(
    title="Proportions of Teams by Tournament Seed",
    width=400,
    height=300
)

plot2
```
